```python
vendas = float (input('Total:R$'))
anos_empresa = int(input('Anos'))
salario_base = 2000.000
comissao = vendas *0.05


if (vendas>10000) and (anos_empresa>=1):
  bonus = 500.00
else:
  bonus = 0.00

novo_salario = salario_base + comissao + bonus
print(f'Novo salario: R${novo_salario:9.2f}')
print(f'Comissao: R${comissao:9.2f}')
print(f'bonus:{bonus:9.2f}')
print(f'Salario final: R${novo_salario:9.2f}')

```

```python
altura_pe_direito= float(input('digite a altura do pé direito da casa :'))
altura_degrau =float(input('digite a altura de cada degraus:'))

quantidade_degraus =altura_pe_direito//altura_degrau
resto = altura_pe_direito % altura_degrau
if resto>0:
  quantidade_degraus +=1
print(f'quantidade de degraus: {quantidade_degraus}')
```

```python
estado_origem=input('digite o estado de produção do produto:')
valor_produto = float(input('digite o valor do produto:'))

if estado_origem =='RJ'and valor_produto>5.000:
  percentual_imposto=0.15
else:
  percentual_imposto=0.05

valor_imposto= valor_produto * percentual_imposto
valor_total= valor_produto + valor_imposto
print(f'valor do imposto:R${valor_imposto:,.2f}')
print(f'digite o valor total a pagar :R${valor_total:,.2f}')
```
