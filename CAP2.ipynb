```python
taxa_kbps = 128 # kilobits por segundo
tempo_min = 3 # duração em minutos
taxa_bps = taxa_kbps * 1000 # bits por segundo
tempo_seg = tempo_min * 60 # duração em segundos
total_bits = taxa_bps * tempo_seg
total_bytes = total_bits / 8
total_mb = total_bytes / 1_000_000
print(f'Total de bits: {total_bits:,.0f}')
print(f'Total de bytes: {total_bytes:,.0f}')
print(f'Tamanho aproximado: {total_mb:.2f} MB')
```
