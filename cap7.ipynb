```python
preco_original = float(input('digite o preço do produto(R$): '))
desconto_porcentagem = float(input('digite a porcentagem de desconto(%): '))

valor_desconto= preco_original * (desconto_porcentagem /100)
preco_final = preco_original - valor_desconto

print(f'desconto aplicado: R${valor_desconto:.2f}')
print(f'preço final com desconto : R$ {preco_final:.2f}')
```
