```python
nota = 0
soma = 0
qtd = 0
while nota != -1 :
  nota = float(input('digite a nota do aluno: '))
  if nota != -1:
    soma = soma+ nota
    qtd = qtd + 1
media = soma / qtd
print(f'média:{media:6.2f}')
```
