### O que é uma String em Python?

Uma **string** (ou cadeia de caracteres) é uma sequência de caracteres usada para representar texto. Em Python, qualquer texto entre aspas simples (`'...'`) ou aspas duplas (`"..."`) é considerado uma string.

No exemplo anterior, você definiu:
```python
frase = "python é fácil"
```

Aqui, a variável `frase` guarda o texto `"python é fácil"`.

#### Operações Comuns com Strings:
1. **Acessar caracteres individuais (Indexação):** O primeiro caractere está na posição `0`.
2. **Tamanho da string:** Usando a função `len()`.
3. **Métodos úteis:**
   - `.upper()`: Transforma tudo em maiúsculas.
   - `.lower()`: Transforma tudo em minúsculas.
   - `.capitalize()`: Deixa a primeira letra maiúscula.
   - `.replace(antigo, novo)`: Substitui uma parte do texto.

````python
# Vamos praticar com a variável 'frase' definida anteriormente
print("Frase original:", frase)

# 1. Tamanho da string
tamanho = len(frase)
print("Quantidade de caracteres:", tamanho)

# 2. Transformando em letras maiúsculas
print("Em maiúsculas:", frase.upper())

# 3. Substituindo uma palavra
nova_frase = frase.replace("fácil", "incrível")
print("Nova frase:", nova_frase)

# 4. Acessando a primeira letra
print("Primeira letra:", frase[0])
````
