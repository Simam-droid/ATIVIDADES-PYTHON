```python
# Exemplo de declaração e atribuição de variáveis

# Atribuindo um número inteiro
idade = 30
print(f"Idade: {idade}, Tipo: {type(idade)}")

# Atribuindo uma string
nome = "Alice"
print(f"Nome: {nome}, Tipo: {type(nome)}")

# Atribuindo um número flutuante (decimal)
preco = 19.99
print(f"Preço: {preco}, Tipo: {type(preco)}")

# Atribuindo um valor booleano
is_ativo = True
print(f"Está ativo: {is_ativo}, Tipo: {type(is_ativo)}")

# Reatribuindo um valor (o tipo pode mudar)
valor = 100
print(f"Valor inicial: {valor}, Tipo: {type(valor)}")

valor = "cem"
print(f"Valor reatribuído: {valor}, Tipo: {type(valor)}")
```
